```python
from ldparser import *
import pandas as pd
import time
import plotly.express as px
```

```python
ld = ldData.fromfile('ld_data/laguna_seca_test.ld')
df = pd.DataFrame(data={c: ld[c].data for c in ld})
print(df)
for col in df.columns:
    if df[col].nunique() == 1 or col not in ['Car Coord X', 'Car Coord Y']:
        df.drop(columns=[col], inplace=True)
# print(df)
```

```python
df.to_csv("ac_coords.csv")
fig = px.scatter(df, x="Car Coord X", y="Car Coord Y")
fig
```

```python
sensors = pd.read_csv('railway_public_sensors.csv')
sensors.drop(columns=['car', 'group', 'type'], inplace=True)
sensor_dict = {}
for idx, row in sensors.iterrows():
    sensor_dict.update({row['name']: row['sensor_id']})

print(sensor_dict)
```

```python
df.rename(columns=sensor_dict, inplace=True)
print(df)
```
